```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

image = cv2.imread("img2.jpeg")

if image is None:
    raise FileNotFoundError("Could not load img2.jpeg")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_float = image.astype(np.float32) / 255.0

plt.figure(figsize=(7, 5))
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
sigma = 2
amount = 1.5

blurred = cv2.GaussianBlur(
    image_float, (0, 0), sigmaX=sigma, sigmaY=sigma
)

sharpened = np.clip(
    image_float + amount * (image_float - blurred), 0, 1
)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(image)
ax[0].set_title("Original Image")

ax[1].imshow(blurred)
ax[1].set_title("Blurred Image")

ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()
```

```python
kernel = np.ones((7, 7), dtype=np.float32) / 49

box_filtered = cv2.filter2D(image_float, -1, kernel)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(np.clip(box_filtered, 0, 1))
plt.title("7x7 Box Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
gaussian_5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21 = cv2.GaussianBlur(image, (21, 21), 0)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(image)
ax[0].set_title("Original Image")

ax[1].imshow(gaussian_5)
ax[1].set_title("Gaussian 5x5")

ax[2].imshow(gaussian_21)
ax[2].set_title("Gaussian 21x21")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()
```

```python
image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

laplacian_display = np.clip(
    (laplacian - laplacian.min()) /
    (laplacian.max() - laplacian.min() + 1e-8),
    0, 1
)

laplacian_sharpened = np.clip(image_float - laplacian, 0, 1)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(image)
ax[0].set_title("Original Image")

ax[1].imshow(laplacian_display)
ax[1].set_title("Laplacian")

ax[2].imshow(laplacian_sharpened)
ax[2].set_title("Laplacian Sharpening")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()
```

```python
#q1

box_3 = cv2.blur(image, (3, 3))
box_9 = cv2.blur(image, (9, 9))
box_15 = cv2.blur(image, (15, 15))

fig, ax = plt.subplots(1, 4, figsize=(20, 6))

ax[0].imshow(image)
ax[0].set_title("Original")

ax[1].imshow(box_3)
ax[1].set_title("Box 3x3")

ax[2].imshow(box_9)
ax[2].set_title("Box 9x9")

ax[3].imshow(box_15)
ax[3].set_title("Box 15x15")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()

```

As the filter size increases, the image becomes more blurred and small details are lost.

```python
#q2

np.random.seed(42)

noisy_image = image.copy()

noise_mask = np.random.random(image.shape[:2])

noisy_image[noise_mask < 0.04] = 0
noisy_image[noise_mask > 0.96] = 255

median_filtered = cv2.medianBlur(noisy_image, 5)
gaussian_filtered = cv2.GaussianBlur(noisy_image, (5, 5), 0)

fig, ax = plt.subplots(1, 4, figsize=(20, 6))

ax[0].imshow(image)
ax[0].set_title("Original")

ax[1].imshow(noisy_image)
ax[1].set_title("Salt-and-Pepper Noise")

ax[2].imshow(median_filtered)
ax[2].set_title("Median Filter 5x5")

ax[3].imshow(gaussian_filtered)
ax[3].set_title("Gaussian Filter 5x5")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()
```

The median filter removes salt-and-pepper noise better than the Gaussian filter while keeping the edges clearer.

```python
#q3
bilateral_filtered = cv2.bilateralFilter(
    image, d=9, sigmaColor=75, sigmaSpace=75
)

gaussian_comparison = cv2.GaussianBlur(image, (9, 9), 0)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))

ax[0].imshow(image)
ax[0].set_title("Original Image")

ax[1].imshow(gaussian_comparison)
ax[1].set_title("Gaussian Filter")

ax[2].imshow(bilateral_filtered)
ax[2].set_title("Bilateral Filter")

for a in ax:
    a.axis("off")

plt.tight_layout()
plt.show()
```

The bilateral filter reduces noise and keeps the edges sharp. It preserves edges better than the Gaussian filter.
